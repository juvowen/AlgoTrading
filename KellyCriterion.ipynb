```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
df = yf.download(['AAPL','NFLX','GOOG'], start='2020-01-01',end='2025-09-09')
df.dropna(inplace=True)
df
```

```python
Returns = df['Close'].pct_change().dropna()
Returns
```

```python
excess_returns = Returns - (0.04/252) #returns over risk-free rate
excess_returns
```

```python
M = 252 * excess_returns.mean()
M
```

```python
C = 252*excess_returns.cov()
C
```

```python
F = np.dot(np.linalg.inv(C),M)
F

#.88X leverag on APPL
#1.11X on google
##1.44x leverage on NLFX
```

```python
S = np.sqrt(np.dot(F.T,np.dot(C,F)))
S
#sharp ratio
```
